## Bài 2.4 — Dự đoán kết quả

Hàm đệ quy dưới đây trả về cặp `(giá trị, số lời gọi)`. Dự đoán kết quả của `g(2468)`, `g(0)` và `g(1999)`. Hàm `g` thực chất tính đại lượng gì của $n$?

```python
def g(n):
    if n == 0:
        return 0, 1
    v, c = g(n // 10)
    return n % 10 + v, c + 1

print(g(2468), g(0), g(1999))
```

## TH 2.4

Nhập liên tục các số thực (bỏ qua và báo giá trị không phải số) cho đến khi gõ `xong`. In số giá trị hợp lệ, tổng, trung bình, giá trị lớn nhất và nhỏ nhất; nếu chưa nhập số nào thì báo ``chưa có dữ liệu''. Kiểm tra với dãy gõ: `7.5, 8, abc, 6.5, 9, xong`.

```python
# TODO: viết lời giải của bạn ở đây
ds = []

while True:
    nhap = input("Nhập số thực (gõ 'xong' để kết thúc): ")

    if nhap.lower() == "xong":
        break

    try:
        so = float(nhap)
        ds.append(so)

    except ValueError:
        print("Lỗi: Giá trị nhập vào không phải số!")

# Thống kê kết quả
if len(ds) == 0:
    print("Chưa có dữ liệu")
else:
    tong = sum(ds)
    trung_binh = tong / len(ds)
    lon_nhat = max(ds)
    nho_nhat = min(ds)

    print(f"Số giá trị hợp lệ: {len(ds)}")
    print(f"Tổng: {tong}")
    print(f"Trung bình: {trung_binh}")
    print(f"Lớn nhất: {lon_nhat}")
    print(f"Nhỏ nhất: {nho_nhat}")
```

## TH 2.7

Dãy Collatz: từ $n$, nếu $n$ chẵn thì $n\leftarrow n/2$, nếu lẻ thì $n\leftarrow3n+1$, lặp đến khi $n=1$. Với $n=27$, đếm số bước và tìm giá trị lớn nhất mà dãy đạt được (đáp án biết trước: $111$ bước, giá trị lớn nhất $9232$).

```python
# TODO: viết lời giải của bạn ở đây
n = 27
so_buoc = 0
gia_tri_lon_nhat = n

while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1

    so_buoc += 1

    if n > gia_tri_lon_nhat:
        gia_tri_lon_nhat = n

print("Số bước:", so_buoc)
print("Giá trị lớn nhất:", gia_tri_lon_nhat)
```

## TH 2.10 ∗

Mở rộng Ví dụ \ref{vd:montepi}: ước lượng $\pi$ với $N=10^2,\ 10^4,\ 10^6$ điểm (dùng `random.seed(2026)` một lần ở đầu). In ước lượng và sai số tuyệt đối; kiểm tra bằng `assert` rằng sai số nhỏ hơn $6/\sqrt N$. Nhận xét: gấp $100$ lần số điểm thì sai số giảm khoảng mấy lần?

```python
# TODO: viết lời giải của bạn ở đây
import random
import math

random.seed(2026)

cac_N = [10**2, 10**4, 10**6]

so_diem_trong = 0
N_max = max(cac_N)

for i in range(1, N_max + 1):
    x = random.random()
    y = random.random()

    if x**2 + y**2 <= 1:
        so_diem_trong += 1

    if i in cac_N:
        pi_uoc_luong = 4 * so_diem_trong / i
        sai_so = abs(pi_uoc_luong - math.pi)
        gioi_han = math.sqrt(6 / i)

        print(f"N = {i}")
        print(f"Pi ước lượng: {pi_uoc_luong:.6f}")
        print(f"Sai số tuyệt đối: {sai_so:.6f}")
        print(f"Giới hạn sai số: {gioi_han:.6f}")

        try:
            assert sai_so < gioi_han
            print("Kiểm tra assert: Đạt")
        except AssertionError:
            print("Kiểm tra assert: Không đạt")

        print()
```

## TH 2.11 ∗

Phương trình $f(x)=x^3-x-2=0$ có đúng một nghiệm trong $[1,2]$ vì $f(1)=-2<0<4=f(2)$ và $f$ tăng trên đoạn này. Cài đặt *phương pháp chia đôi*: giữ đoạn $[a,b]$ mà $f(a)f(b)<0$, thay bằng nửa đoạn chứa nghiệm cho tới khi $b-a<10^{-10}$. In nghiệm, số bước, và so sánh với kết quả lý thuyết: số bước cần thiết là $\lceil\log_2(1/10^{-10})\rceil$.

```python
# TODO: viết lời giải của bạn ở đây
import math

def f(x):
    return x**3 - x - 2

a = 1
b = 2
epsilon = 1e-10
so_buoc = 0

while b - a >= epsilon:
    c = (a + b) / 2

    if f(a) * f(c) < 0:
        b = c
    else:
        a = c

    so_buoc += 1

# Nghiệm gần đúng
nghiem = (a + b) / 2

# Số bước lý thuyết
so_buoc_ly_thuyet = math.ceil(
    math.log2((b - a + epsilon) / epsilon)
)

print(f"Nghiệm gần đúng: {nghiem:.12f}")
print(f"Số bước thực tế: {so_buoc}")
print(f"Số bước lý thuyết: {math.ceil(math.log2(1 / epsilon))}")
```
