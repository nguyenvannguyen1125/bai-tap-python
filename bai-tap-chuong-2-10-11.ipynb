### A.3 Hàm, tham số mặc định, đệ quy

```python
def la_nguyen_to(n: int) -> bool:
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print([n for n in range(2, 31) if la_nguyen_to(n)])

def giai_thua(n: int) -> int:
    return 1 if n <= 1 else n * giai_thua(n - 1)

def fib_dem(n):
    if n < 2:
        return n, 1
    a, ca = fib_dem(n - 1)
    b, cb = fib_dem(n - 2)
    return a + b, ca + cb + 1

print(giai_thua(5), fib_dem(10), fib_dem(20))
assert fib_dem(20) == (6765, 21891)
```

## Bài 2.4 — Dự đoán kết quả

Hàm đệ quy dưới đây trả về cặp `(giá trị, số lời gọi)`. Dự đoán kết quả của `g(2468)`, `g(0)` và `g(1999)`. Hàm `g` thực chất tính đại lượng gì của $n$?

```python
def g(n):
    if n == 0:
        return 0, 1
    v, c = g(n // 10)
    return n % 10 + v, c + 1

print(g(2468), g(0), g(1999))
```

## Bài 2.6 — Đọc lỗi và sửa mã 🤖 (mã do công cụ AI sinh)

Một công cụ AI sinh hàm kiểm tra số nguyên tố dưới đây. Chạy với $n=1,2,9,15,17$, đối chiếu với đáp án đúng lần lượt là `False`, `True`, `False`, `False`, `True`, tìm *ít nhất hai lỗi*, sửa hàm và viết `assert` cho cả năm trường hợp.

```python
def la_nguyen_to(n):
    for i in range(2, n):
        if n % i == 0:
            return False
        else:
            return True

for n in (1, 2, 9, 15, 17):
    print(n, la_nguyen_to(n))
```

## Bài 2.9 — Bẫy logic
