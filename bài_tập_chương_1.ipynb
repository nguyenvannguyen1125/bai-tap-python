```python
x = 23
y = 6

print("x / y =", x / y)
print("x // y =", x // y)
print("x % y =", x % y)
print("x ** 2 =", x ** 2)
print("-x // y =", -x // y)
print("-x % y =", -x % y)
print("2 ** 3 ** 2 =", 2 ** 3 ** 2)
print("(2 ** 3) ** 2 =", (2 ** 3) ** 2)
```

```python
s = "KY THUAT LAP TRINH"

print("len(s) =", len(s))
print("s[0] =", repr(s[0]))
print("s[-1] =", repr(s[-1]))
print("s[3:8] =", repr(s[3:8]))
print("s[:2] =", repr(s[:2]))
print("s[::-1][:4] =", repr(s[::-1][:4]))
print("s[::3] =", repr(s[::3]))
print("s[30:] =", repr(s[30:]))
```

```python
print(round(2.5), round(3.5), round(0.5), round(1.2345, 2))
print(int(-3.7), -3.7 // 1, int(3.99))
print(int("12") + int("3"), "12" + "3")
print("ab" * 3 + "c", 7 / 2 * 2, 10 - 2 ** 2 * 3)
```

### không cùng làm tròn lên vì Python sử dụng quy tắc làm tròn về số nguyên chẵn gần nhất (Round half to even)

```python
a, b = 7, 2
print(a, b, sep="")
print(a / b, a // b, sep=" | ")
print(f"{a}+{b}={a + b}", end="; ")
print(f"{a / b:.1f}")
print("a" "b", 3 * "ab")
print(f"{1234.5678:,.2f}", f"{5:03d}", f"{0.256:.1%}")
```
