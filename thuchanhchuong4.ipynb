```python
import math
class HinhCau:
    def __init__(self, ban_kinh):
        if ban_kinh <= 0:
            raise ValueError("Nhap ban kinh lon hon 0")
        self.ban_kinh = ban_kinh
    def dien_tich(self):
        return 4 * math.pi * self.ban_kinh**2
    def the_tich(self):
        return (4/3) * math.pi * self.ban_kinh**3
    def __str__(self):
        return f"Hình cầu có bán kính = {self.ban_kinh}"
hinh1 = HinhCau(2)
print(hinh1)
print("Diện tích:", hinh1.dien_tich())
print("Thể tích:", hinh1.the_tich())
try:
    hinh2 = HinhCau(-1)
except ValueError as e:
    print("Lỗi:", e)
```

```python
class SoDuKhongDu(Exception):
    pass
class TaiKhoan:
    def __init__(self, chu_tai_khoan, so_du=0):
        if so_du < 0:
            raise ValueError("Số dư ban đầu không được âm")
        self.chu_tai_khoan = chu_tai_khoan
        self.so_du = so_du
        self.lich_su = []
    def nap(self, tien):
        if tien <= 0:
            raise ValueError("Số tiền nạp phải lớn hơn 0")
        self.so_du += tien
        self.lich_su.append(f"Nạp: +{tien}")
        assert self.so_du >= 0
        return self.so_du
    def rut(self, tien):
        if tien <= 0:
            raise ValueError("Số tiền rút phải lớn hơn 0")
        if tien > self.so_du:
            raise SoDuKhongDu("Số dư không đủ để thực hiện giao dịch")
        self.so_du -= tien
        self.lich_su.append(f"Rút: -{tien}")
        assert self.so_du >= 0
        return self.so_du
    def __str__(self):
        return f"Chủ tài khoản: {self.chu_tai_khoan}, Số dư: {self.so_du}"

tk = TaiKhoan("Nguyen Van A", 1000000)
print("Ban đầu:", tk)
tk.nap(500000)
print("Sau khi nạp:", tk.so_du)
tk.rut(200000)
print("Sau khi rút:", tk.so_du)
try:
    tk.rut(2000000)
except SoDuKhongDu as e:
    print("Lỗi:", e)
for giao_dich in tk.lich_su:
    print(giao_dich)
print("\nSố dư cuối cùng:", tk.so_du)
assert tk.so_du >= 0
```

```python
import math
class HinhHoc:
    def dien_tich(self):
        raise NotImplementedError("Lớp con phải cài đặt phương thức dien_tich()")
    def __str__(self):
        return "Hình học"
class HinhTron(HinhHoc):
    def __init__(self, r):
        self.r = r
    def dien_tich(self):
        return math.pi * self.r**2
    def __str__(self):
        return f"Hình tròn (r={self.r})"
class HinhChuNhat(HinhHoc):
    def __init__(self, a, b):
        self.a = a
        self.b = b
    def dien_tich(self):
        return self.a * self.b
    def __str__(self):
        return f"Hình chữ nhật ({self.a} x {self.b})"
class HinhTamGiac(HinhHoc):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def dien_tich(self):
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(p * (p - self.a) * (p - self.b) * (p - self.c))
    def __str__(self):
        return f"Hình tam giác ({self.a}, {self.b}, {self.c})"
ds_hinh = [
    HinhTron(1),
    HinhChuNhat(3, 4),
    HinhTamGiac(3, 4, 5)
]
ds_hinh.sort(key=lambda h: h.dien_tich())
for h in ds_hinh:
    print(f"{h}: diện tích = {h.dien_tich():.3f}")
tong = sum(h.dien_tich() for h in ds_hinh)
print("\nTổng diện tích:", tong)
print("Giá trị kiểm tra π + 18:", math.pi + 18)
assert math.isclose(tong, math.pi + 18)
```

```python
import pandas as pd
df = pd.read_csv("goldprice.csv")
df["Date"] = pd.to_datetime(
    df["Date"], format="%Y-%m", errors="coerce"
)
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
df = df.dropna(subset=["Date", "Price"])

df["Year"] = df["Date"].dt.year

yearly = df.groupby("Year").agg(
    gia_trung_binh=("Price", "mean"),
    so_thang=("Date", "nunique")
)
yearly = yearly[yearly["so_thang"] == 12].copy()
yearly["tang_giam_%"] = [
    (yearly.loc[y, "gia_trung_binh"]
     / yearly.loc[y - 1, "gia_trung_binh"] - 1) * 100
    if y - 1 in yearly.index else float("nan")
    for y in yearly.index
]
result = yearly.dropna(subset=["tang_giam_%"])
top_3 = result.sort_values(
    "tang_giam_%", ascending=False
).head(3)

print(top_3[["gia_trung_binh", "tang_giam_%"]].round(2))
giam_manh_nhat = result.sort_values(
    "tang_giam_%", ascending=True
).head(1)

print(giam_manh_nhat[["gia_trung_binh", "tang_giam_%"]].round(2))
```

```python
import csv
with open("luong.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["nam_kn", "luong"])
    writer.writerows([
        [1, 30],
        [2, 35],
        ["abc", 40],
        [3, ""],
        [4, 50],
        [5, 55]
    ])

class DongKhongHopLe(Exception):
    pass

def doc_du_lieu(ten_file):
    dong_hop_le = []
    dong_loi = []
    tong_luong = 0

    with open(ten_file, "r", encoding="utf-8") as f:
        reader = csv.DictReader(f)

        for so_dong, row in enumerate(reader, start=2):
            try:
                nam_kn = int(row["nam_kn"])
                luong = float(row["luong"])

                dong_hop_le.append({
                    "nam_kn": nam_kn,
                    "luong": luong
                })

                tong_luong += luong

            except (ValueError, TypeError, KeyError):
                try:
                    raise DongKhongHopLe(
                        f"Dòng {so_dong} không hợp lệ"
                    )
                except DongKhongHopLe:
                    dong_loi.append(so_dong)

    trung_binh = tong_luong / len(dong_hop_le) if dong_hop_le else 0

    return dong_hop_le, dong_loi, trung_binh

hop_le, loi, trung_binh = doc_du_lieu("luong.csv")
for row in hop_le:
    print(row)
print("\nDanh sách dòng lỗi:", loi)
print("Lương trung bình:", trung_binh)
```
