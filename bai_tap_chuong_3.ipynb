```python
# 1. Khai báo dữ liệu bằng dict
data = { 1: 4.0, 2: 5.5, 3: 6.0, 4: 7.5, 5: 8.0 }

x = list(data.keys())
y = list(data.values())

# 2. Tính trung bình
x_tb = statistics.mean(x)
y_tb = statistics.mean(y)

# 3. Tính hệ số b
tu_so = sum((xi - x_tb) * (yi - y_tb) for xi, yi in zip(x, y))
mau_so = sum((xi - x_tb) ** 2 for xi in x)

b = tu_so / mau_so

# 4. Tính hệ số a
a = y_tb - b * x_tb

print("Hệ số a:", a)
print("Hệ số b:", b)
print(f"Phương trình hồi quy: y = {a} + {b}x")

# 5. Dự đoán khi ôn tập 6 giờ
x_moi = 6
y_moi = a + b * x_moi

print("Điểm dự đoán:", y_moi)

# 6. Kiểm tra bằng statistics.linear_regression
import statistics
import matplotlib.pyplot as plt

result = statistics.linear_regression(x, y)

print("Kiểm tra bằng thư viện:")
print("Slope:", result.slope)
print("Intercept:", result.intercept)
```

```python
from collections import defaultdict

# 1. Khai báo danh sách sinh viên
students = [("An", 9.0), ("Binh", 7.5), ("Chi", 8.6), ("Dung", 5.0), ("Em", 6.0), ("Phuc", 3.5), ("Giang", 7.0)]

# 2. Hàm quy đổi điểm chữ
def xep_loai(diem):
    if diem >= 8.5:
        return "A"
    elif diem >= 7:
        return "B"
    elif diem >= 5.5:
        return "C"
    elif diem >= 4:
        return "D"
    else:
        return "F"

# 3. Tạo từ điển để gom nhóm
groups = defaultdict(list)

for ten, diem in students:
    diem_chu = xep_loai(diem)
    groups[diem_chu].append(ten)

# 4. In kết quả theo thứ tự chữ cái
for nhom in sorted(groups):
    print(f"Điểm {nhom}: {groups[nhom]}")
```

```python
# 1. Khai báo doanh thu hai cửa hàng
store1 = { "ao": 120, "quan": 80, "non": 30 }

store2 = { "ao": 90, "giay": 150, "non": 20 }

# 2. Hợp nhất doanh thu theo sản phẩm
total = {}

for sp in store1.keys() | store2.keys():
    total[sp] = store1.get(sp, 0) + store2.get(sp, 0)

# 3. In doanh thu theo thứ tự giảm dần
print("Tổng doanh thu theo sản phẩm:")
for sp, dt in sorted(total.items(), key=lambda x: x[1], reverse=True):
    print(sp, ":", dt)

# 4. Tìm sản phẩm bán ở cả hai cửa hàng
both = store1.keys() & store2.keys()
print("\nSản phẩm bán ở cả hai cửa hàng:", both)

# 5. Tìm sản phẩm chỉ bán ở một cửa hàng
only_one = store1.keys() ^ store2.keys()
print("Sản phẩm chỉ bán ở một cửa hàng:", only_one)
```

```python
# 1. Khởi tạo tập hợp các số từ 2 đến 1000
n = 1000
S = set(range(2, n + 1))

# 2. Sàng Eratosthenes
for p in range(2, int(n**0.5) + 1):
    if p in S:
        # Loại bỏ các bội số của p
        S.difference_update(range(p * p, n + 1, p))

# 3. Sắp xếp các số nguyên tố
primes = sorted(S)

# 4. Tính số lượng và tổng
so_luong = len(primes)
tong = sum(primes)

# 5. Đếm số cặp số nguyên tố sinh đôi
twin_primes = [(p, p + 2) for p in S if p + 2 in S]

# 6. In kết quả
print("Các số nguyên tố:", primes)
print("Số lượng số nguyên tố:", so_luong)
print("Tổng các số nguyên tố:", tong)
print("Các cặp số nguyên tố sinh đôi:", sorted(twin_primes))
print("Số cặp số nguyên tố sinh đôi:", len(twin_primes))
```
